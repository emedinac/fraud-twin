# Getting Started with FraudTwin

FraudTwin can be used directly as a Python library. This tutorial generates a small, deterministic payment world entirely in memory.

## 1. Install FraudTwin

Install FraudTwin in your Python environment before importing it. The command below is shown for reference and is intentionally not executed by this tutorial.

```bash
python -m pip install fraudtwin
```

If FraudTwin is already imported, restart the kernel after installation so the notebook loads the newly installed version.

### Optional: Neo4j support

When connecting to a Neo4j database, install the official Neo4j Python Driver separately:

```bash
python -m pip install neo4j
```

See the [Neo4j Python Driver Manual](https://neo4j.com/docs/python-manual/current/) for the [installation guide](https://neo4j.com/docs/python-manual/current/install/) and connection examples.

## 2. Generate a payment world

The default is in-memory generation, so no temporary directories or command-line wrappers are needed.

````python
import fraudtwin

data = fraudtwin.generate()

print(data.run_id)
````

## 3. Inspect the generated data

The result contains reproducibility metadata and typed Python objects. You can inspect counts without reading files or calling the CLI.

````python
print(data.manifest.entity_counts)
print(data.manifest.event_counts)
````

````python
for payment in data.behavior.payments[:3]:
    print(
        {
            "payment_id": payment.payment_id,
            "rail": payment.payment_rail,
            "amount": payment.amount,
            "status": payment.current_status,
        }
    )
````

## 4. Use the ML dataset

The minimal configuration also builds point-in-time ML rows. The returned dataset can be used directly as a Polars DataFrame.

````python
data.dataset.frame.head()
````

## 5. Count generated cases

These counts are payment cases: a payment linked to a fraud record is counted as fraud, and the remaining payments are counted as non-fraud. The ML dataset may contain fewer point-in-time rows, and its labels can be unavailable when no fraud workflow is enabled.

````python
total_cases = len(data.behavior.payments)
fraud_payment_ids = {record.payment_id for record in data.behavior.fraud_records}
fraud_cases = len(fraud_payment_ids)
non_fraud_cases = total_cases - fraud_cases
ml_points = data.dataset.frame.height if data.dataset is not None else 0

print(f"Payment cases: {total_cases}")
print(f"Fraud cases: {fraud_cases}")
print(f"Non-fraud cases: {non_fraud_cases}")
print(f"ML dataset points: {ml_points}")
````

That is the complete basic workflow: install the package, generate data, inspect the returned Python objects, and understand the generated case counts.
